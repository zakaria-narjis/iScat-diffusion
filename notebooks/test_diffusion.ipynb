```python
timesteps = 1000
num_steps = 250

step_size = timesteps // num_steps
timesteps = list(range(0, timesteps, step_size))
timesteps = list(reversed(timesteps))  # Start from highest timestep
print(timesteps)
```

```python
import torch
timesteps = 1000
num_steps = 250
timesteps = torch.linspace(
    timesteps-1, 0, num_steps, dtype=torch.long
).tolist() 
print(timesteps)
```
